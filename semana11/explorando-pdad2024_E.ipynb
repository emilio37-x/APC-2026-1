```python
import pandas as pd

# Forcing Pandas to use calamine instead of openpyxl
domicilios = pd.read_excel("data/domicilios.xlsx", engine="calamine")

for _, linha in domicilios.iterrows():
    print(f"Domicílio {linha['A01nficha']}: {linha['A01npessoas']} moradores, {linha['A01ncriancas']} crianças")
```
